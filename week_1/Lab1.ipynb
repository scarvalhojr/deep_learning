### Tensorflow: Accurate Digit Classification in 14 Lines

```python
# ================================
# Environment Setup and Library Versions
# ================================

# Import system module to check Python version
import sys

# Import TensorFlow library for building and training neural networks
import tensorflow as tf

# Print Python version
print("Python version:", sys.version)

# Print TensorFlow version
print("TensorFlow version:", tf.__version__)
```

```python
# ================================
# MNIST Handwritten Digit Classification using TensorFlow
# ================================

# Step 1: Load the MNIST dataset
# The MNIST dataset contains 70,000 images of handwritten digits (0–9).
# It is split into training and test sets.
mnist = tf.keras.datasets.mnist

# 60,000 training samples and 10,000 test samples,
(x_train, y_train),(x_test, y_test) = mnist.load_data()

print("x_train shape: %", x_train.shape)

# Step 2: Normalize the data
x_train = tf.keras.utils.normalize(x_train, axis=1)
x_test = tf.keras.utils.normalize(x_test, axis=1)

# Step 3: Create a Sequential neural network
model = tf.keras.models.Sequential()

# Step 4: Add layers
model.add(tf.keras.layers.Flatten())
model.add(tf.keras.layers.Dense(64, activation=tf.nn.sigmoid))
model.add(tf.keras.layers.Dense(128, activation=tf.nn.sigmoid))
model.add(tf.keras.layers.Dense(10, activation=tf.nn.softmax))

# Step 5: Compile the model
model.compile(optimizer="adam", loss="sparse_categorical_crossentropy", metrics=["accuracy"])

# Step 6: Train the model
model.fit(x=x_train, y=y_train, epochs=5)

# Step 7: Display model summary
print(model.summary())

# Step 8: Evaluate model performance on test data
test_loss, test_acc = model.evaluate(x=x_test, y=y_test, verbose=0)

# Step 9: Print test accuracy as a percentage
print('\nTest accuracy: % `%`', test_acc*100)
```

The code above trains a Deep Neural Network (DNN) to classify handwritten images like the one below to digits. Using modern optimizers and normalized data, it can reach around 96% accuracy in just 5 training epochs. Thanks to the Keras interface in TensorFlow, and assuming your data is organized correctly, you can easily adapt this model to new problems.

## Appendix B - Linear Algebra in Python

In Python, we can use the `numpy` library to easily define and perform operations on vectors and matrices.

### Matrices
The term **matrix** refers to a 2D rectangular array of numbers.


```python
import numpy as np

# Define a matrix from a string, using semicolons to separate rows
A = np.matrix('1 2 3; 4 5 6')
print("Matrix A (from string):\n", A)

# Define a matrix from a list of lists, where each inner list is a row
B = np.matrix([[1, 2, 3], [4, 5, 6]])
print("\nMatrix B (from list of lists):\n", B)

# Using numpy's array constructor to create a 2D array
C = np.array([[1, 2, 3], [4, 5, 6]])
print("\nArray C (2D numpy array):\n", C)
```

A matrix has a specific dimensionality, defined by its number of rows and columns. Using standard mathematical notation, a matrix with 2 rows and 3 columns of real numbers can be represented as $R^{3x2}$.

In Python, the shape attribute or function in NumPy can be used to determine the number of rows and columns of a matrix. For example, matrix.shape will return a tuple (num_rows, num_columns).

```python
print(A.shape)
```

We refer to individual objects within the matrix as **elements**. We use the subscript notation on the matrix  in order to refer to individual objects. In general $M_{i,j}$ will refer to the element at the $i^{th}$ row and $j^{th}$ column of M.

NumPy supports a wide range of methods for indexing and slicing arrays, which we will not cover in detail here. In simple cases, however, we can use indices to directly access and manipulate elements of a matrix, as shown below.

```python
print(A[1,1])
```

Note that indexing for rows and columns in NumPy matrices starts at 0.

### Vectors
A vector is a 1-dimensional (1D) array and can be considered a special case of a matrix with only one column.

Although vectors are mathematically a type of matrix, in practice we often use different operators and functions to create and manipulate vectors in NumPy.

```python
# Create a NumPy vector using a standard Python list
v1 = np.array([2, 3, 1, 0])
print(v1)

# Note: this is NOT equivalent to the vector above.
# Here we are creating a 1x4 matrix (a single row matrix),
# not a 1D vector.
v2 = np.matrix([[2, 3, 1, 0]])
print(v2)

# Index an element in the vector (indexing starts at 0)
print(v1[1])
```

### Working with Matrices and Vectors

> In linear algebra, operations on matrices and vectors are only valid when their dimensions are compatible.

We can add and subtract matrices (or vectors) **only if they have the same dimensionality**.  

When this condition is met, the result is a new matrix (or vector) with the **same shape** as the inputs.

Addition and subtraction are performed **element-wise**. That is, each element in one matrix is added to or subtracted from the corresponding element in the other matrix.

```python
# Add two matrices A and B element-wise
# Both matrices must have the same shape
C = A + B
print("Matrix C (A + B):")
print(C)

# Subtract matrix A from matrix C element-wise
D = C - A
print("\nMatrix D (C - A):")
print(D)
```

We can do the same with the vectors:

```python
# Define a second vector with the same dimensionality as v1
v2 = np.array([10, 20, 30, 40])

# Add two vectors element-wise
# Each element in v1 is added to the corresponding element in v2
v3 = v1 + v2
print("v1 + v2 =", v3)

# Subtract v1 from v2 element-wise
# Each element in v1 is subtracted from the corresponding element in v2
v3 = v2 - v1
print("v2 - v1 =", v3)
```

We can also directly apply scalar multiplication and division to matrices and vectors. In these operations, each element of the matrix or vector is multiplied or divided by the same scalar value.



```python
# Scalar multiplication on a matrix
# Each element of A is multiplied by 2
E = A * 2
print(E)

# Scalar division on a matrix
# Each element of A is divided by 2
F = A / 2
print(F)

# Scalar multiplication on a vector
# Each element of v2 is multiplied by 2
v4 = v2 * 2
print(v4)

# Scalar division on a vector
# Each element of v2 is divided by 2
v5 = v2 / 2
print(v5)
```

### Inner / Scalar Product

The inner product (also called the scalar product or dot product) of two vectors is a single scalar value obtained by multiplying corresponding elements of the vectors and summing the results. This operation is defined only for vectors of equal length.

In geometric terms, the inner product is related to the angle and similarity between two vectors. Rather than directly measuring distance, it is often used as a measure of similarity between vectors. For example, it is widely used in applications such as document similarity, recommendation systems, and clustering.

In Python, we can compute the inner product of two vectors using NumPy’s inner function.

```python
# Compute the inner (dot) product of two vectors
dot_product = np.inner(v1, v2)
print("Inner product of v1 and v2:", dot_product)
```

Note that the `dot` can also be used with matrices and higher-dimensional arrays.

### Matrix-Matrix Multiplication
Given two matrices, we can calculate their matrix product (also called the matrix-matrix multiplication) if the number of columns in the first matrix equals the number of rows in the second matrix.

For two matrices A and B, the dimensionality of the resultant matrix product is given by:

\begin{equation}
R_{A}C_{A} \times R_{B}C_{B} = R_{A}C_{B}
\end{equation}

The operation for calculating the matrix product is straightforward: the entry for row i column j in the resultant matrix C is the dot product of the i$^{th}$ row of A and the j$^{th}$ column of B.

In python we can use the NumPy function **matmul** to perform matrix-matrix multiplication.

```python
G = np.matrix('1 2; 3, 4; 5, 6')
print(G)
H = np.matmul(A,G)
print(H)
```

Remember that matrix-matrix multiplication is not commutative.
\begin{equation}
  A \times B \neq B \times A
\end{equation}

```python
A = np.matrix('1, 2, 3; 4, 5, 6; 7, 8, 9')
B = np.matrix('1, 1, 1; 2, 2, 2; 3, 3, 3')
print(A)
print(B)
print(np.matmul(A,B))
print(np.matmul(B,A))
```

### Matrix Identity and Inverse

We can define the identity $I$ of a matrix which in general allows the commutation property to hold:

\begin{equation}
 A \times I = I \times A = A
\end{equation}

Here $I$ is the Identity Matrix which is a square matrix, and all of its diagonal elements are equal to 1 and all of its non-diagonal elements are equal to 0.

NumPy allows us to easily define an identity matrix with a specified number of rows and columns.

```python
I = np.identity(3)
print(I)
print(A)
print(np.matmul(A,I))
print(np.matmul(I,A))
```

Just as we can define the inverse for a real number $X \in R$ as $\frac{1}{X}$, we can also define the inverse for a matrix.

Beginning first with the case of real numbers, we note that:

\begin{equation}
 X \times INV(X) = I
\end{equation}

where I is the identity for real numbers - which is 1.

This gives us an intuition of how the inverse is defined for matrices:  

\begin{equation}
 A \times INV(A) = I
\end{equation}

i.e., the inverse of a matrix A should be defined such that, the matrix product of A with itself should produce an identity matrix.

While we can straightforwardly calculate the inverse of a real number as $\frac{1}{X}$, computing the inverse of a matrix is more complicated and involves determinants and cofactors, which we will not cover here. Fortunately, we can compute the inverse directly in NumPy.

```python
import numpy as np
import numpy.linalg as la

# Define a 3x3 matrix X
X = np.matrix('100, -200, 403; 44, -5, 607; 27, 98, -59')

# Compute the inverse of X
B = la.inv(X)

# Print the inverse
print("Inverse of X (B):")
print(B)

# Verify by multiplying X and its inverse (should give identity matrix)
print("\nX @ B (should be close to identity):")
print(np.matmul(X, B))

# Note: A singular matrix is a square matrix that does not have an inverse.
# Inverting matrices that are near-singular (determinant close to 0)
# can lead to large numerical errors or failures
print("\nUncomment lines in the block below to test with a near-singular matrix A (expect an error)!")
# B = la.inv(A)
# print(np.matmul(A, B))
```

### Matrix Transpose
We can also define the transpose of a matrix $A^{T}$ as the matrix obtained by swapping the rows and columns of A.

\begin{equation}
   A_{ij} = A_{ji}
\end{equation}

```python
# Compute the transpose of matrix A
C = A.T

# Print the original matrix
print("Original matrix A:")
print(A)

# Print the transposed matrix
print("\nTransposed matrix C = A.T:")
print(C)
```
